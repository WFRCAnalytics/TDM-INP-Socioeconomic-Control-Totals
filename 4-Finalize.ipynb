```python
import pandas as pd
import numpy as np
```

```python
se_df = pd.concat([pd.read_csv('intermediate/split_emp.csv'), pd.read_csv('intermediate/split_pop.csv')])
se_df
```

```python
se_df[(se_df['YEAR']==2023) & (se_df['VARIABLE']=='HH')]
```

```python
se_pivot_df = se_df.pivot(index=['CO_FIPS','YEAR'],columns='VARIABLE',values='VALUE').reset_index().fillna(0).astype(int)
se_pivot_df
```

```python
se_final_df = se_pivot_df.copy()

se_final_df[';INDEX'] = se_final_df['CO_FIPS'].astype(str) + se_final_df['YEAR'].astype(str)

se_final_df['HHSIZE'] = se_final_df['HHPOP'] / se_final_df['HH']
se_final_df['HHSIZE'] = se_final_df['HHSIZE'].replace([np.inf, -np.inf], 0)
se_final_df['HHSIZE'] = se_final_df['HHSIZE'].fillna(0).round(2)

se_final_df['ALLEMP'] = se_final_df['FOOD'] + se_final_df['MANU'] + se_final_df['WSLE'] + se_final_df['OFFI'] + se_final_df['GVED'] + se_final_df['HLTH'] + se_final_df['OTHR'] + se_final_df['AGRI'] + se_final_df['MING'] + se_final_df['CONS'] + se_final_df['HBJ']

se_final_df = se_final_df[[';INDEX','CO_FIPS','YEAR','TOTPOP','GQPOP','HHPOP','HH','HHSIZE','POP0017','POP1864','POP65P','ALLEMP','RETL','FOOD','MANU','WSLE','OFFI','GVED','HLTH','OTHR','AGRI','MING','CONS','HBJ']]
se_final_df
```
